```python
import apple
```

```python
asin_str='''
{ asin ← refl. (λz. {
  as ⟜ ⟨1.5707288,_0.2121144,0.0742610,_0.0187293⟩;
  Ho ← gen. 1 (z*) 4;
  𝜋%2-√(1-z)*Ho⋅as
});
  (asin')
}
'''
```

```python
asin=apple.jit(asin_str)
```

```python
import numpy as np
```

```python
xs=np.arange(-1,1,0.01,dtype=np.float64)
```

```python
ys=asin(xs)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(ys)
```

```python
plt.plot(np.asin(xs))
```
