```python
import apple
```

```python
bessel_str='''
λ𝛼.λx.
  { fact ← [(*)/ₒ 1 (⍳ 1 x)]
  ; f10 ← λa.λx.
    { rf ← [(*)/ₒ 1 (frange x (x+y-1) (⌊y))]; ffact ← rf 1
    ; mkIx ← λk. {kk⟜ℝ k; (x^k%(rf a kk))%ffact kk}
    ; (+)/(mkIx'(⍳ 0 30))
    }
  ; ((x%2)^𝛼%ℝ(fact 𝛼))*f10 (ℝ(𝛼+1)) (_((x^2)%4))
  }
'''
```

```python
bessel=apple.jit(bessel_str)
```

```python
import numpy as np
```

```python
xs=np.arange(0,20,0.1,dtype=np.float64)
```

```python
ys=np.vectorize(lambda x: bessel(1,x))(xs)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(ys)
```
