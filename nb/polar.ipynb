```python
import apple
```

```python
import matplotlib.pyplot as plt
```

```python
# Lemniscate of Bernoulli: https://mathworld.wolfram.com/Lemniscate.html
```

```python
lemni=apple.jit("λN. {ts ⟜ frange 0 (2*𝜋) N; ((λt. cos.t%(1+sin.t^2))'ts, (λt. sin. t*cos. t%(1+sin. t^2))'ts)}")
```

```python
lemni_x,lemni_y=lemni(200);plt.plot(lemni_x,lemni_y)
```

```python
# Aztema spiral https://www.2dcurves.com/spiral/spiralat.html
```

```python
aztema=apple.jit("λN. {ts ⟜ frange 1 (1+2*𝜋) N; ((λt. sin.t%t-2*cos.t-t*sin. t)'ts,(λt. cos.t%t-2*sin.t+t*cos. t)'ts)}")
```

```python
aztema_x,aztema_y=aztema(200);plt.plot(aztema_x,aztema_y)
```

```python
# https://mathworld.wolfram.com/DevilsCurve.html
```

```python
devil=apple.jit('''
λN. 
  { ts ⟜ (frange 0 (2*𝜋) N)
  ; ((λt. cos.t*√((sin. t^2-2*cos. t^2)%(sin. t^2-cos. t^2)))'ts,(λt. sin.t*√((sin. t^2-2*cos. t^2)%(sin. t^2-cos. t^2)))'ts)
  }
''')
```

```python
devil_x,devil_y=devil(700);plt.plot(devil_x,devil_y)
```
