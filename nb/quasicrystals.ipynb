```python
import matplotlib.pyplot as plt
```

```python
import apple
```

```python
v=apple.jit("λN.{k ⟜ 2*𝜋%ℝN; ([cos.(k*x)]'(frange 0 (ℝN-1) N), [sin.(k*x)]'(frange 0 (ℝN-1) N))}")
```

```python
step=apple.jit('''
λxs.
{ φ ← (1+√5)%2
; ♭(xs [φ^2*x-φ*y]⊗ xs)
}''')
```

```python
v_x,v_y=v(5)
```

```python
N=3
v_x,v_y=v(5)
for _ in range(N):
    v_x = step(v_x)
    v_y = step(v_y)
```

```python
def make_plot():
    fig, ax = plt.subplots(figsize=(6,6))
    fig.set_facecolor("k")
    ax.set_facecolor("k")
    ax.scatter(v_x, v_y, s=1, c="w")
    ax.axis("off")
    return fig
```

```python
make_plot()
```
